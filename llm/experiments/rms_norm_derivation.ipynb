```python
x = torch.tensor([1, 2, 3])
d = x.shape[0]
rms_x = torch.sqrt((x ** 2).sum(dim=0) / d)
y = x / rms_x
mag_y = torch.sqrt((y ** 2).sum(dim=0))
root_d = torch.sqrt(torch.tensor(d))
rms_y = torch.sqrt((y ** 2).sum(dim=0) / d)

print(f'{mag_y=}\n{root_d=}\n{mag_y == root_d=}\n{rms_y=}')
```

```python
weight = 2
rms_y_scaled_by_weight = torch.sqrt(((y * weight) ** 2).sum(dim=0) / d)
print(f'{rms_y_scaled_by_weight}')
```
