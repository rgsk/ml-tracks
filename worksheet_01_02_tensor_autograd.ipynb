## Question 7 — `cat` vs `stack`

**TASK:**
```text
x = torch.tensor([[1., 2.], [3., 4.]])
y = torch.tensor([[5., 6.], [7., 8.]])
```
Compute `c0 = torch.cat([x, y], dim=0)`, `c1 = torch.cat([x, y], dim=1)`, `s0 = torch.stack([x, y], dim=0)`. Print each and `.shape`.

**PREDICT:** Shapes of `c0`, `c1`, `s0`? In one sentence: how is stacking different from concatenating here?

````python
# YOUR CODE HERE
from tkinter import Y


x= torch.tensor([[1., 2.], [3., 4.]])
y= torch.tensor([[5., 6.], [7., 8.]])

c0= torch.cat([x, y], dim=1)
print("c0:", c0)
s0_0= torch.stack([x, y], dim=0) 
print("s0 stacked dim 0:", s0_0)

#for each row, you get a row from x and a row from y
s0_1= torch.stack([x, y], dim=1) 
print("s0 stacked dim 1:", s0_1)

#for each col, you get a col from x and a col from y
s0_2= torch.stack([x, y], dim=2) 
print("s0 stacked dim 2:", s0_2)

#stack inserts a new axis at position dim and stacks x and y along that axis
#cat makes it one long vector
# c0=[2,2] c1=[2,2] s0=[2,2,2]
````

## Question 8 — Scalar autograd

**TASK:**
```text
x = torch.tensor([3.0], requires_grad=True)
y = torch.tensor([2.0], requires_grad=True)
z = x * y + x ** 2
```
Run `z.backward()` and print `x.grad` and `y.grad`.

**PREDICT:** `dz/dx` = ? , `dz/dy` = ? (calculus)

````python
# YOUR CODE HERE
x= torch.tensor([3.0], requires_grad= True)
y= torch.tensor([2.0], requires_grad= True)
z= x* y+x **2
z.backward()
print("dz/dx:", x.grad) #predict its 8
print("dz/dy:", y.grad) #predict its 3
````

## Question 9 — Vector output & `backward`

**TASK:** `x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)`, `y = x ** 2`. **Without** `y.backward(torch.tensor([1.,1.,1.]))`, get the same `x.grad` by reducing `y` to a scalar first, then `.backward()`. Print `x.grad` (should match `2 * x` element-wise).

**PREDICT:** Why does `backward()` need a scalar (or an explicit gradient argument)?

````python
# YOUR CODE HERE
x= torch.tensor([1.0, 2.0, 3.0], requires_grad= True)
y= x**2
y.sum().backward() #use sum when you're plugging in a vector into a derivative
print("dy/dx", x.grad)
````

## Question 10 — Matrix mean gradient + NumPy bridge

**TASK (autograd):** `x = torch.ones(2, 2) * 2.0` with `requires_grad=True`. `z = (x ** 2).mean()`. Print `z` (predict `4.0`). `z.backward()`; print `x.grad`. **PREDICT:** every `x.grad[i,j]` = ?

**TASK (NumPy):** `arr = np.array([1., 2., 3., 4.])`, `torch.from_numpy(arr)`, print tensor. Change `arr[0]` in NumPy; print tensor again (shared memory).

````python
# YOUR CODE HERE

import torch
import numpy as np


x= torch.full((2, 2), 2.0, requires_grad= True)
print(x)
z= (x**2).mean()
print(z.item())
z.backward()
print("dy/dx:", x.grad)

arr= np.array([1., 2., 3., 4.])
print("tensor", torch.from_numpy(arr))
arr[0]= 6
print("tensor_changed", torch.from_numpy(arr))
````
