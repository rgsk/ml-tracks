- **`view()`** only works on **contiguous tensors** (tensors stored sequentially in memory).
- **`transpose()`** changes how data is interpreted **without physically rearranging memory**, making the tensor **non-contiguous**.
- Calling `view()` on a non-contiguous tensor throws an error.
- **`reshape()`** is more flexible because it may internally create a copy if required.
- **`.contiguous()`** creates a new tensor with rearranged memory, allowing `view()` to work again.


```python
import torch

x = torch.arange(12)
print(f"{x=}")
y = x.view(3, 4)
print(f"{y=}")
print(f"{y.shape=}, {y.stride()=}")
print(f"{y.view(12)=}")
y_t = y.transpose(0, 1)
print(f"y_t:\n{y_t}")
print(f"{y_t.shape=}, {y_t.stride()=}")
try:
    print(f"{y_t.view(12)=}")
except Exception as e:
    print(f"Error: {e}")
print(f"{y_t.reshape(12)=}")
print(f"{y_t.contiguous().view(12)=}")
print(f"y_t.contiguous():\n{y_t.contiguous()}")
print(f"{y_t.contiguous().shape=}, {y_t.contiguous().stride()=}")
```
